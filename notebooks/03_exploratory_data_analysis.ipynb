# 03 - Analyse exploratoire

Analyse des variables et de la cible `Churn`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
import streamlit as st

print("Toutes les bibliothèques sont installées.")
```

```python
# Répartition des clients selon Churn
df = pd.read_csv('D:\\IBM\\data\\IBM-Telco-Customer-Churn.csv') 
plt.figure(figsize=(6,4))
sns.countplot(x='Churn', data=df)
plt.title('Répartition des clients selon leur statut de départ')
plt.xlabel('Statut de départ')
plt.ylabel('Nombre de clients')
plt.savefig(
    "D:/IBM/reports/figures/churn_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Churn selon le genre
plt.figure(figsize=(6,4))
sns.countplot(x='Churn', hue='gender', data=df)
plt.title('Répartition des clients selon leur statut de départ et le genre')
plt.xlabel('Statut de départ')
plt.ylabel('Nombre de clients')
plt.savefig(
    "D:/IBM/reports/figures/churm_by_gender.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Churn selon le contrat
plt.figure(figsize=(6,4))
sns.countplot(x='Churn', hue='Contract', data=df)
plt.title('Répartition des clients selon leur statut de départ et le type de contrat')
plt.xlabel('Statut de départ')
plt.ylabel('Nombre de clients')
plt.savefig(
    "D:/IBM/reports/figures/churn_by_contract.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
#Churn selon l'ancienneté
plt.figure(figsize=(6,4))
sns.boxplot(x='Churn', y='tenure', data=df)
plt.title('Répartition des clients selon leur statut de départ et leur ancienneté')
plt.xlabel('Statut de départ')
plt.ylabel('Nombre de clients')
plt.savefig(
    "D:/IBM/reports/figures/churm_by_tenure.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Churn selon les frais mensuels
plt.figure(figsize=(6,4))
sns.boxplot(x='Churn', y='MonthlyCharges', data=df)
plt.title('Répartition des clients selon leur statut de départ et les frais mensuels')
plt.xlabel('Statut de départ')
plt.ylabel('Frais mensuels')
plt.savefig(
    "D:/IBM/reports/figures/churn_by_MonthlyCharges.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Churn selon le paiement
plt.figure(figsize=(6,4))
sns.barplot(x='Churn', hue='PaymentMethod', data=df)
plt.title('Répartition des clients selon leur statut de départ et le mode de paiement')
plt.xlabel('Statut de départ')
plt.ylabel('Nombre de clients')
plt.savefig(
    "D:/IBM/reports/figures/churn_by_payment.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Churn selon Internet
plt.figure(figsize=(6,4))
sns.countplot(x='Churn', hue='InternetService', data=df)
plt.title('Répartition des clients selon leur statut de départ et le type de service Internet')
plt.xlabel('Statut de départ')
plt.ylabel('Nombre de clients')
plt.savefig(
    "reports/figures/churn_by_InternetService.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Churn selon TechSupport
plt.figure(figsize=(6,4))
sns.barplot(x='Churn', hue='TechSupport', data=df)
plt.title('Répartition des clients selon leur statut de départ et le support technique')
plt.xlabel('Statut de départ')
plt.ylabel('Nombre de clients')
plt.savefig(
    "D:/IBM/reports/figures/churn_by_TechSupport.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Distribution TotalCharges
plt.figure(figsize=(6,4))
sns.histplot(df['TotalCharges'], bins=30, kde=True)
plt.title('Distribution des frais totaux')
plt.xlabel('Frais totaux')
plt.ylabel('Nombre de clients')
plt.savefig(
    "D:/IBM/reports/figures/churn_by_TotalCharges.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Corrélations numériques

# Sélection unique des colonnes numériques
df_numeric = df.select_dtypes(include=['number'])

# Créer la figure
plt.figure(figsize=(10, 8))

sns.heatmap(df_numeric.corr(),annot=True,cmap='coolwarm',fmt='.2f',linewidths=0.5)

plt.title('Matrice de corrélation des variables numériques')
plt.savefig(
    "D:/IBM/reports/figures/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
