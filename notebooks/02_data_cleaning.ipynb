```python
df["TotalCharges"] =pd.to_numeric(df['TotalCharges'], errors='coerce')

df["TotalCharges"] 
```

```python
numeric_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

df[numeric_columns].describe()
```

```python
# Recherche des valeurs aberrantes
for column in numeric_columns:

    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\nVariable : {column}")
    print("Limite inférieure :", lower_bound)
    print("Limite supérieure :", upper_bound)
    print("Nombre d'outliers :", len(outliers))
```

```python
# Vérifier les valeurs négatives
for column in numeric_columns:
    print(f"{column} : {(df[column] < 0).sum()} valeurs négatives")
```

### Vérification des valeurs aberrantes et des valeurs négatives

La méthode IQR a été utilisée pour détecter les valeurs aberrantes dans les variables `tenure`, `MonthlyCharges` et `TotalCharges`.

Aucune valeur aberrante n'a été détectée.

Une vérification complémentaire a également été effectuée pour rechercher les valeurs négatives. Les résultats montrent :
- `tenure` : 0 valeur négative.
- `MonthlyCharges` : 0 valeur négative.
- `TotalCharges` : 0 valeur négative.

**Conclusion :** les trois variables numériques ne présentent aucune anomalie détectée par ces contrôles. Aucune observation n'a été supprimée.
