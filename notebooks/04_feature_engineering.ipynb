Préparation des variables pour l'entraînement du modèle.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
import streamlit as st

print("Toutes les bibliothèques sont installées.")
```

```python
df = pd.read_csv('D:\\IBM\\data\\IBM-Telco-Customer-Churn_cleaned.csv')  # Remplacez 'data.csv' par le chemin de votre fichier de données
```

```python
#Identifier les variables intéressantes.
categorical_columns = [
    'gender',
    'Partner',
    'Dependents',
    'PhoneService',
    'InternetService',
    'OnlineSecurity',
    'TechSupport',
    'Contract',
    'PaperlessBilling',
    'PaymentMethod'
]
```

```python

for column in categorical_columns:
    churn_rate = pd.crosstab( df[column], df['Churn'],normalize='index') * 100

    print(f"\nTaux de résiliation selon {column}")
    print(churn_rate['Yes'])
```

### Analyse des variables catégorielles et du taux de résiliation

L'analyse des taux de résiliation selon les variables catégorielles révèle plusieurs différences intéressantes.

- **Gender :** les taux de résiliation sont très proches entre les femmes (26,92 %) et les hommes (26,16 %). Cette variable semble peu discriminante.
- **Partner :** les clients sans partenaire présentent un taux de résiliation plus élevé (32,96 %) que ceux ayant un partenaire (19,66 %).
- **Dependents :** les clients sans personnes à charge présentent un taux de résiliation de 31,28 %, contre 15,45 % pour ceux qui en ont.
- **PhoneService :** les taux sont relativement proches entre les deux catégories.
- **PaymentMethod :** le paiement par chèque électronique présente un taux de résiliation particulièrement élevé (45,29 %), tandis que la carte bancaire automatique présente un taux de 15,24 %.

**Conclusion :** PaymentMethod, Partner et Dependents constituent des variables intéressantes à approfondir dans la suite de l'analyse exploratoire. Ces observations devront être complétées par l'étude des autres variables et confirmées lors de la modélisation.

```python
#analyser Contract
contract_churn_rate = pd.crosstab(df['Contract'], df['Churn'], normalize='index') * 100
print("\nTaux de résiliation selon Contract")
print(contract_churn_rate['Yes'])
```

### Analyse du taux de résiliation selon le type de contrat

L'analyse de la variable `Contract` révèle une différence importante entre les catégories.

- **Month-to-month :** 42,71 % de résiliation.
- **One year :** 11,27 % de résiliation.
- **Two year :** 2,83 % de résiliation.

Les clients ayant un contrat mensuel présentent le taux de résiliation le plus élevé, tandis que ceux ayant un contrat de deux ans présentent le taux le plus faible.

Cette observation suggère que la durée d'engagement contractuel est fortement associée au comportement de résiliation des clients.

**Conclusion :** la variable `Contract` constitue une caractéristique prioritaire à conserver et à étudier lors de la modélisation prédictive.

```python
#analyser les variables numériques
for column in ['tenure', 'MonthlyCharges', 'TotalCharges']:

    print(f"\nAnalyse de {column}")

    print(
        df.groupby('Churn')[column].mean()
    )
```

```python
numeric_columns = [
    'tenure',
    'MonthlyCharges',
    'TotalCharges'
]

for column in numeric_columns:

    plt.figure(figsize=(5, 3))

    sns.boxplot(
        data=df,
        x='Churn',
        y=column,
        hue='Churn',
        palette={'Yes': 'red', 'No': 'green'},
        legend=False
    )

    plt.title(f'{column} selon le statut de résiliation')
    plt.xlabel('Statut de résiliation')
    plt.ylabel(column)

    plt.show()
```

### Analyse des variables numériques selon le statut de résiliation

L'analyse des moyennes des variables numériques révèle plusieurs différences entre les clients fidèles et les clients ayant résilié leur abonnement.

**1. Tenure :**
- Clients fidèles : 37,57 mois.
- Clients résiliés : 17,98 mois.

Les clients qui résilient présentent une ancienneté moyenne nettement plus faible.

**2. MonthlyCharges :**
- Clients fidèles : 61,27.
- Clients résiliés : 74,44.

Les clients ayant résilié leur abonnement paient en moyenne davantage chaque mois.

**3. TotalCharges :**
- Clients fidèles : 2 555,34.
- Clients résiliés : 1 531,80.

Les clients ayant résilié présentent des dépenses cumulées moyennes plus faibles. Cette différence peut notamment être liée à leur ancienneté plus courte.

**Conclusion :** les variables `tenure`, `MonthlyCharges` et `TotalCharges` présentent des différences intéressantes selon le statut de résiliation. Elles seront conservées pour la suite de l'analyse et évaluées lors de la modélisation.
